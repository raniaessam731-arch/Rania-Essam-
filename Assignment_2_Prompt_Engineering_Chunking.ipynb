## Task 1 – Prompt Engineering


```python
prompts = [
    'Explain Cloud Computing to a beginner.',
    'Explain Cloud Computing to a beginner in simple English. Do not use technical jargon, do not mention specific cloud providers, and do not exceed 50 words.',
    'Explain Cloud Computing using a real-life example in 60–70 words. Do not use acronyms, technical terms, bullet points, or complicated explanations.'
]

results = [
    'Cloud Computing means using computer services through the internet instead of relying only on your own device.',
    'Cloud Computing means using computer services through the internet. Instead of storing everything on your own computer, you can use online resources to store files, run applications, and access information from different devices.',
    'Cloud Computing is like renting a storage room instead of keeping everything in your home. You can store your files online and access them whenever you need them. For example, when you save photos online, you can open them from your phone or computer without keeping all the files on one device.'
]

for i, (p, r) in enumerate(zip(prompts, results), 1):
    print(f'PROMPT {i}:\n{p}\nRESULT {i}:\n{r}\n' + '-'*70)
```

```python
assert len(results) == 3
assert all('Cloud Computing' in r for r in results)
assert len(results[1].split()) <= 50
assert 50 <= len(results[2].split()) <= 80
print('Task 1 tests passed.')
```

### Observation
Prompt 1 gives a general answer. Prompt 2 reduces jargon and unnecessary details. Prompt 3 adds stronger constraints for length, style, and format, producing the most controlled response.


## Task 2 – Chunking Strategy


```python
text = '''## Databases
A database is an organized collection of information that can be stored and managed efficiently.

## SQL
SQL is a language used to communicate with databases. It allows users to retrieve, insert, update, and delete data.

## Database Security
Database security protects stored information from unauthorized access and prevents data from being changed or deleted without permission.
'''

def fixed_size_chunks(text, chunk_size=100, overlap=20):
    chunks=[]
    start=0
    while start < len(text):
        end=start+chunk_size
        chunks.append(text[start:end].strip())
        start=end-overlap
    return [c for c in chunks if c]

import re
def section_chunks(text):
    sections=re.split(r'(?=## )', text)
    return [s.strip() for s in sections if s.strip()]

fixed_chunks=fixed_size_chunks(text)
semantic_chunks=section_chunks(text)
print('Fixed-size chunks:', len(fixed_chunks))
print('Section-based chunks:', len(semantic_chunks))
for i,c in enumerate(semantic_chunks,1): print(f'Chunk {i}: {c}\n')
```

```python
query='What can SQL be used for?'
sql_chunk=next(c for c in semantic_chunks if c.startswith('## SQL'))
print('Retrieved chunk:\n', sql_chunk)
assert 'retrieve' in sql_chunk.lower()
assert 'insert' in sql_chunk.lower()
assert 'update' in sql_chunk.lower()
assert 'delete' in sql_chunk.lower()
print('Task 2 retrieval test passed.')
```

### Observation
Fixed-size chunking can split related sentences. Section-based chunking keeps each topic and its supporting sentences together, so the complete SQL section is retrieved for the query.

## Final Conclusion
Negative constraints improve control over generated responses, while section-based chunking preserves topic-level context and improves retrieval completeness.
